```python
import ast

# Checking if 116 unique events in athletes.csv is correct (some have "" as their event name)
athlete_events = set()
empty_event_count = 0

for e in df_athletes["events"]:
    events_list = ast.literal_eval(e)   # convert string -> list of dicts
    for event in events_list:
        athlete_events.add((event["discipline"], event["event"]))

empty_events = [pair for pair in athlete_events if pair[1] == ""]

print("Unique events in athletes.csv:", len(athlete_events))
print("Empty events in athletes.csv:", len(empty_events))
```

Conclusion:

Used this large table to understand where the 6 too many events come from

Some events are named with training:

-Nordic Combined	Large Hill Training instead of Nordic Combined	Individual Gundersen Large Hill/10km

-Nordic Combined	Normal Hill Training instedad of Nordic Combined	Individual Gundersen Normal Hill/10km

Also some events just have different nomenclature names:

-Men's LH Individual vs. Men's Large Hill

-Men's NH Individual vs. Men's Normal Hill

-Women's LH Individual vs. Women's Large Hill

-Women's NH Individual vs. Women's Normal Hill

```python
df_schedules["event"] = df_schedules["event"].replace("Large Hill Training", "Individual Gundersen Large Hill/10km")
df_schedules["event"] = df_schedules["event"].replace("Normal Hill Training", "Individual Gundersen Normal Hill/10km")
df_schedules["event"] = df_schedules["event"].replace("Men's Large Hill", "Men's LH Individual")
df_schedules["event"] = df_schedules["event"].replace("Men's Normal Hill", "Men's NH Individual")
df_schedules["event"] = df_schedules["event"].replace("Women's Large Hill", "Women's LH Individual")
df_schedules["event"] = df_schedules["event"].replace("Women's Normal Hill", "Women's NH Individual")
df_schedules.groupby(["discipline", "event"]).ngroups
```

```python
disciplines_set = set()

for disciplines in df_venues["disciplines"]:
    # Ignore NaN
    if pd.isna(disciplines):
        continue
    
    try:
        disciplines_list = ast.literal_eval(disciplines)   # convert string -> list (for multiple disciplines at venue otherwise crash)
        disciplines_set.update(disciplines_list)

    except (ValueError, SyntaxError):
        # If ast.literal_eval fails (case for single disciplines at venue), add the value directly
        disciplines_set.add(disciplines)

print("Total unique disciplines: " + str(len(disciplines_set)))
print(disciplines_set)
```

```python
# Doing the same for disciplines_code
disciplines_set = set()

for disciplines in df_venues["disciplines_code"]:
    # Ignore NaN
    if pd.isna(disciplines):
        continue
    
    try:
        disciplines_list = ast.literal_eval(disciplines)   # convert string -> list (for multiple disciplines at venue otherwise crash)
        disciplines_set.update(disciplines_list)

    except (ValueError, SyntaxError):
        # If ast.literal_eval fails (case for single disciplines at venue), add the value directly
        disciplines_set.add(disciplines)

print("Total unique disciplines: " + str(len(disciplines_set)))
print(disciplines_set)
```
