# Quality check

## Duplicates

```python
duplicated_rows = df[df.duplicated()]
duplicated_rows
```

```python
df_unique = df.drop_duplicates()
df_unique.shape
```

```python
df_unique.nunique()
```

## Borough distribution

```python
df_unique.groupby("Boro").size()
```

## Time distribution

```python
df_unique.groupby("RequestID").size().mean()
```

```python
rec_per_day = df_unique.groupby(["Yr", "M", "D"]).size()
rec_per_day = rec_per_day.reset_index(name="Cnt")
rec_per_day
```

```python
rec_per_day["date"] = pd.to_datetime(rec_per_day["Yr"].astype(str) + "-" + rec_per_day["M"].astype(str) + "-" + rec_per_day["D"].astype(str))
rec_per_day = rec_per_day.sort_values("date") # Looks like already the case but let's make sure
rec_per_day
```

```python
import plotly.express as px

fig = px.line(
    rec_per_day,
    x="date",
    y="Cnt",
    title="Traffic recordings per day",
    labels={"date": "Date", "Cnt": "Number of recordings"}
)

fig.update_traces(
    hovertemplate="Date: %{x|%Y-%m-%d}<br>Recordings: %{y}<extra></extra>"
)

fig.show()
```

```python
all_days = pd.date_range("2010-01-01", "2020-12-31", freq="D")
missing_days = set(all_days.date) - set(rec_per_day["date"].dt.date)
len(missing_days)
```

```python
len(missing_days) / len(all_days) * 100
```

```python
missing_series = pd.Series(pd.to_datetime(list(missing_days)))
missing_per_year = missing_series.dt.year.value_counts().sort_index()
missing_per_year
```

## Time of the day distribution (per year)

```python
df_unique["minute_of_day"] = df_unique["HH"] * 60 + df_unique["MM"]
df_unique.head()
```

```python
dist_day = (
    df_unique
    .groupby(["Yr", "minute_of_day"])
    .size()
    .reset_index(name="Cnt")
)
dist_day
```

```python
fig = px.line(
    dist_day,
    x="minute_of_day",
    y="Cnt",
    color="Yr",
    title="Distribution of recordings during the day by year",
    labels={
        "minute_of_day": "Minute of day",
        "Cnt": "Number of recordings",
        "Yr": "Year"
    }
)

fig.show()
```
